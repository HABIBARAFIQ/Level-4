## Define Function for using numpy

1. forward propagation,
2. loss -> Mean Squared Error(MSE)->MSE = 1/N * (w*x − y) \*\*2
3. gradient -> dJ/dw = 1/N _ 2x _ (w\*x -y) [element wise multiply]


```python
def forward(x):
    return weights*x
```

```python
def loss(y_predicted,y):
    return ((y_predicted-y)**2).mean()
```

```python
def gradient(x, y, y_predicted):
    return np.multiply(2 * x, (y_predicted - y)).mean()
```

### Let's say we are predicting for f(x) = 2x


```python
X = np.array([1, 2, 3, 4],dtype=np.float32)
Y = np.array([2, 4, 6, 8],dtype=np.float32)
X,Y
```

```python
learning_rate = 0.01
weights = 0
epochs = 100
```

```python
print(f'Prediction before training: {forward(5):.3f}')
```

```python
for epoch in range(epochs):
    # step-1 :forward pass
    y_pred = forward(X)
    # step-2 :loss calculation
    l = loss(y_pred,Y)
    # step-3 :Calculate gradient
    dw = gradient(X,Y,y_pred)
    #step-4 : weights update
    weights -= learning_rate * dw
    print(f'epoch:{epoch} y-pred:{y_pred} loss:{l} w:{weights}')
```

```python
print(f'Prediction after training: {forward(5):.3f}')
```

## Define Function for using torch

1. forward propagation,
2. loss -> Mean Squared Error(MSE)->MSE = 1/N * (w*x − y) \*\*2
3. gradient -> dJ/dw = 1/N _ 2x _ (w\*x -y) [element wise multiply] -> Back propagation


```python
X = torch.tensor([1, 2, 3, 4],dtype=torch.float32)
Y = torch.tensor([2, 4, 6, 8],dtype=torch.float32)
X,Y
```

```python
learning_rate = 0.01
weights = torch.tensor(0.0, requires_grad = True)
epochs = 100
```

```python
print(f'Prediction before training: {forward(5):.3f}')
```

```python
for epoch in range(epochs):
    # step-1 :forward pass
    y_pred = forward(X)
    # step-2 :loss calculation
    l = loss(y_pred,Y)
    # step-3 :Calculate gradient ->now do backpropagate
    l.backward() #dl/dw
    #step-4 : weights update
    with torch.no_grad():
        weights -= learning_rate * weights.grad
    weights.grad.zero_()    
    print(f'epoch:{epoch} y-pred:{y_pred} loss:{l} w:{weights}')
```

```python
print(f'Prediction after training: {forward(5):.3f}')
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn import datasets
import torch
from sklearn.model_selection import train_test_split
```

```python
bc = datasets.load_breast_cancer()
X, y = bc.data, bc.target
n_samples, n_features = X.shape
n_out_samples = y.size
n_out_features = 1
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.2, random_state=1234)
```

```python
scale = StandardScaler()
X_train = scale.fit_transform(X_train)
X_test = scale.transform(X_test)
```

```python
X_train = torch.from_numpy (X_train.astype(np.float32))
X_test = torch.from_numpy (X_test.astype(np.float32))
y_train = torch.from_numpy (y_train.astype(np.float32))
y_test = torch.from_numpy (y_test.astype(np.float32))  
```

```python
y_train = y_train.view(y_train.shape[0],1)
y_test = y_test.view(y_test.shape[0],1)
```

## Defining custom Logistic Regression model


```python
class LogisticRegression(nn.Module):
    def __init__(self,input_dim,output_dim):
        super(LogisticRegression,self).__init__()
        self.linear = nn.Linear(input_dim,output_dim)
    def forward(self,x):
        y_predicted = torch.sigmoid(self.linear(x)) 
        return y_predicted    
        
```

```python
model = LogisticRegression(n_features,n_out_features)
```

```python
## define loss and optimizer
criterion = nn.BCELoss()
optimizer = torch.optim.SGD(model.parameters(),lr=0.01)
```

```python
for epoch in range(epochs):
    y_predicted = model(X_train)
    l = criterion(y_predicted,y_train)
    l.backward()
    optimizer.step()
    optimizer.zero_grad()
    print(f'epoch:{epoch} loss:{l} ')
```

```python
with torch.no_grad():
    y_predicted = model(X_test)
    y_predicted_cls = y_predicted.round()
    acc = y_predicted_cls.eq(y_test).sum() / float(y_test.shape[0])
    print(f'accuracy: {acc:.4f}')
```
